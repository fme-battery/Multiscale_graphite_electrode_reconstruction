```python
# Load your surface mesh from an STL file
surface_mesh = trimesh.load('Particle_Prism.stl')

# Convert the surface mesh to a solid mesh
solid_mesh = surface_mesh.convex_hull

# Save the solid mesh as an STL file
solid_mesh.export('Particle_Prism_Solidified.stl')
```

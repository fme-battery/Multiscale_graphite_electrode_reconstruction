```python
nii_file = nib.load('Particle_Prism.nii')
```

```python
# Get the header
header = nii_file.header

# Change the pixel dimension
header['pixdim'][1:4] = header['pixdim'][1:4] / 14

# Change the space unit
current_units = header['xyzt_units']
new_units = (current_units & 0xF8) | 0x03  # Preserve temporal units
header['xyzt_units'] = new_units
```

```python
# Save the modified NIfTI file
nib.save(nii_file, 'Particle_Prism_updated.nii')
```
